## 📋 CAMPOS CLAVE POR TABLA 

customers 

customer_id, first_name, last_name, gender, age, 
city, country, registration_date, loyalty_tier 
(Bronze / Silver / Gold / VIP), 
acquisition_channel (store / online / event / referral), 
email_opt_in, preferred_store_id 
  

stores 

store_id, store_name, city, country, store_type 
(flagship / corner / outlet / online), 
opening_date, size_sqm 
  

products 

product_id, product_name, category 
(bags / shoes / ready-to-wear / accessories / fragrances), 
subcategory, unit_cost, unit_price, brand_line 
(main / diffusion / exclusive), season 
  

transactions 

transaction_id, customer_id, store_id, transaction_date, 
total_amount, discount_applied, payment_method, 
channel (in-store / online / phone), campaign_id (nullable) 
  

transaction_items 

item_id, transaction_id, product_id, 
quantity, unit_price, line_total 
  

(Esta tabla es la que habilita UPT, sell-through y análisis de categorías) 

campaigns 

campaign_id, campaign_name, campaign_type 
(email / SMS / event / push / direct mail), 
start_date, end_date, target_segment, budget, objective 
(reactivation / upsell / loyalty / acquisition) 
  

campaign_responses 

response_id, campaign_id, customer_id, 
sent_date, opened (bool), clicked (bool), 
converted (bool), revenue_attributed 
  

returns 

return_id, transaction_id, customer_id, product_id, 
return_date, quantity_returned, refund_amount, 
return_reason (size / defect / preference / gift / other), 
return_channel (in-store / online / courier), 
refund_method (original_payment / store_credit), 
days_since_purchase 
  

(Esta tabla habilita Return Rate por producto/canal/razón, Net Revenue real y análisis de calidad)

## 🏗️ ARQUITECTURA DE BASE DE DATOS — DIAGRAMA ERD INICIAL 

Este diagrama representa las relaciones entre las 8 tablas del modelo. Las claves primarias (PK) y foráneas (FK) definidas aquí se usarán directamente en el DDL SQL. El diagrama es una primera versión — puede evolucionar al construir el DDL. 

 

customers (PK: customer_id) 
  |-- transactions (FK: customer_id, store_id, campaign_id) 
        |-- transaction_items (FK: transaction_id, product_id) 
        |-- returns (FK: transaction_id, customer_id, product_id) 
  |-- campaign_responses (FK: customer_id, campaign_id) 
stores (PK: store_id) 
  |-- transactions (FK: store_id) 
products (PK: product_id) 
  |-- transaction_items (FK: product_id) 
  |-- returns (FK: product_id) 
campaigns (PK: campaign_id) 
  |-- transactions (FK: campaign_id, nullable) 
  |-- campaign_responses (FK: campaign_id) 

 

Ver diagrama visual completo en el archivo ERD adjunto (generado con dbdiagram.io / draw.io). Este bloque de texto sirve como referencia rápida de relaciones para el DDL.
